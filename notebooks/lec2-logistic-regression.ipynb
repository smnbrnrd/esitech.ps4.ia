We give below two functions to help you test your implementation. The first function generates a toy dataset with two classes and the second function plots the decision boundaries of a classifier on this dataset.

```python
from matplotlib import pyplot as plt
from sklearn.datasets import make_blobs

def generate_toydataset(n_train=200):
    mus = [[-1.0, -1.0], [1.0, 1.0]]
    sigmas = [0.8, 0.8]
    X, y = make_blobs(n_samples=n_train, centers=mus, cluster_std=sigmas, n_features=2, random_state=42)
    return X, y

X, y = generate_toydataset()
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm')
```

```python
def plot_decision_frontiers(clf, X, y, ax, h = 0.02):
    x_min, x_max = X[:, 0].min() - .5, X[:, 0].max() + .5
    y_min, y_max = X[:, 1].min() - .5, X[:, 1].max() + .5
    xx, yy = np.meshgrid(np.arange(x_min, x_max, h), np.arange(y_min, y_max, h))
    xtest_grid = np.c_[xx.ravel(), yy.ravel()]
    Z = clf.predict_proba(xtest_grid)[:, 1]
    Z = Z.reshape(xx.shape)
    ax.contourf(xx, yy, Z, cmap='coolwarm', alpha=.6)
    ax.scatter(X[:, 0], X[:, 1], c=y, cmap='coolwarm')
    
h_ps4 = LogisticRegressionPS4(max_iterations=200, learning_rate=0.1, debug=True).fit(X, y)
_, ax = plt.subplots()
plot_decision_frontiers(h_ps4, X, y, ax)
```
